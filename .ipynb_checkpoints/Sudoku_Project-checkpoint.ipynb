## 1. Introduction

Sudoku is a logic-based number puzzle.

The goal is to fill a 9×9 grid with digits from
1 to 9 so that each digit appears exactly once
in every row, every column, and every 3×3 box.

This project explores the mathematical principles
behind Sudoku and implements a Sudoku solver
using Python.

The main solving technique is the backtracking
algorithm.

The project also investigates the execution time
of the algorithm using different Sudoku puzzles.

## 2. Project Objectives

The main objectives of this project are:

1. Explain the mathematical rules of Sudoku.
2. Represent a Sudoku puzzle using Python.
3. Implement a backtracking algorithm.
4. Solve different Sudoku puzzles.
5. Visualize the original and solved boards.
6. Measure the execution time.
7. Analyze the results using basic statistics.
