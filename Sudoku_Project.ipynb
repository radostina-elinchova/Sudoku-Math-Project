## 3. Mathematical Background

### 3.1 Sudoku as a Constraint Satisfaction Problem

Sudoku is a mathematical and logical puzzle that can be modeled as a Constraint Satisfaction Problem (CSP).

A standard Sudoku board consists of 81 cells arranged in a 9 x 9 grid, divided into nine 3 x 3 subgrids [1].

The goal is to fill the empty cells with digits from 1 to 9 while satisfying the following constraints:

1. Each row must contain every digit from 1 to 9 exactly once.
2. Each column must contain every digit from 1 to 9 exactly once.
3. Each 3 x 3 subgrid must contain every digit from 1 to 9 exactly once.
4. The original given digits must remain unchanged.

**Mathematical Model**

A Constraint Satisfaction Problem consists of three main components: variables, domains, and constraints [2].

**1. Variables**

Each cell of the Sudoku board is represented by a variable:

$$
X_{r,c}
$$

where \(r\) represents the row and \(c\) represents the column.

A standard Sudoku puzzle contains 81 such variables.

**2. Domains**

The domain of a variable represents the possible values that can be assigned to it.

For an initially empty cell, the domain is:

$$
D_{r,c} = \{1,2,3,4,5,6,7,8,9\}
$$

For a cell containing an original given digit, the domain contains only that fixed value.

**3. Constraints**

The mathematical constraints determine which combinations of values are allowed.

All variables belonging to the same row must have different values. The same condition applies to columns and 3 x 3 subgrids.

For example, two different cells in the same row must satisfy:

$$
X_{r,c_1} \neq X_{r,c_2}, \quad c_1 \neq c_2
$$

Similar conditions apply to cells belonging to the same column or subgrid.

Peter Norvig [1] describes Sudoku using units and peers. A unit is a row, column, or 3 x 3 subgrid. Peers are other cells that share a unit with a given cell.

**Connection to the Solving Algorithm**

The objective is to assign a valid value to every empty cell while satisfying all mathematical constraints simultaneously.

The backtracking algorithm will use these constraints to reject invalid assignments and search for a complete solution.

This mathematical model provides the foundation for the Sudoku-solving algorithm developed in this project.

The missing digits can be found using set difference:

$$

$$

In Sudoku, this operation allows us to determine which
digits are missing from a particular row.

### 3.4 Finding Candidate Values

Identifying missing digits in a row is only the first
step in solving Sudoku.

A possible value for an empty cell must satisfy three
constraints simultaneously: it must not already appear
in the corresponding row, column, or 3 x 3 subgrid [1].

Let:

- $D$ be the set of allowed digits.
- $R$ be the set of digits used in the row.
- $K$ be the set of digits used in the column.
- $B$ be the set of digits used in the 3 x 3 subgrid.
- $C$ be the set of possible candidate values.

The candidates can be calculated using:

$$
C = D \setminus (R \cup K \cup B)
$$

This expression combines two fundamental operations
from set theory: union and set difference [3].

The union operation combines the digits already used
in the row, column, and subgrid.

The set difference operation removes these digits
from the complete set of allowed values.

**Mathematical Example**

Consider the following sets:

$$
D = \{1,2,3,4,5,6,7,8,9\}
$$

$$
R = \{3,5,7\}
$$

$$
K = \{8\}
$$

$$
B = \{3,5,6,8,9\}
$$

First, we calculate the union:

$$
R \cup K \cup B = \{3,5,6,7,8,9\}
$$

Then, we calculate the set difference:

$$
C = D \setminus (R \cup K \cup B)
$$

$$
C = \{1,2,4\}
$$

Therefore, the possible candidate values are 1, 2,
and 4.

These values satisfy the local row, column, and
subgrid restrictions.

However, a locally valid candidate does not
necessarily lead to a complete Sudoku solution.

The backtracking algorithm will use these candidate
sets to explore possible assignments and return
to previous decisions when a contradiction occurs.

In Python, set union is implemented using `|`,
and set difference using `-`.

This demonstrates how mathematical set operations
can be translated into Python code.

This is one reason why backtracking is useful: the algorithm can try a candidate, continue solving, and return to an earlier decision if a contradiction occurs [1].

## 5. Checking Sudoku Constraints

### 5.1 Mathematical Definition of a Valid Move

Sudoku can be modeled as a Constraint Satisfaction
Problem (CSP), where each cell is associated with
a variable and a domain of possible values [2].

A candidate digit can be placed in an empty cell
only if it does not violate the row, column,
or 3 x 3 subgrid constraints [1].

Let $A$ be the current Sudoku matrix and let
$(i,j)$ be the zero-based coordinates of an
empty cell.

We want to determine whether a digit $v$,
where $v \in D$, can be placed at this position.

The set of allowed digits is:

$$
D = \{1,2,3,4,5,6,7,8,9\}
$$

Define the following sets:

$$
R_i = \{a_{i,k} \mid 0 \leq k \leq 8,\ a_{i,k} \neq 0\}
$$

$$
K_j = \{a_{k,j} \mid 0 \leq k \leq 8,\ a_{k,j} \neq 0\}
$$

$$
B_{i,j} =
\{a_{r,c} \mid
s_r \leq r < s_r+3,\ 
s_c \leq c < s_c+3,\ 
a_{r,c} \neq 0\}
$$

where the starting indices of the subgrid are:

$$
s_r = 3\left\lfloor\frac{i}{3}\right\rfloor
$$

$$
s_c = 3\left\lfloor\frac{j}{3}\right\rfloor
$$

A digit is locally valid if:

$$
v \notin R_i
\quad \land \quad
v \notin K_j
\quad \land \quad
v \notin B_{i,j}
$$

The symbol $\land$ represents logical AND.

Equivalently, the set of candidate values is:

$$
C_{i,j} = D \setminus (R_i \cup K_j \cup B_{i,j})
$$

Therefore, a candidate is locally valid exactly when:

$$
v \in C_{i,j}
$$

These formulas use set operations and logical
conjunction, which are fundamental concepts
of discrete mathematics [3].

A locally valid move satisfies the current
Sudoku constraints, but it does not guarantee
that the entire puzzle can be solved.

This distinction is important for the
backtracking algorithm, which may need
to reconsider earlier assignments [1].

### 5.2 Implementing the Constraint Check

The mathematical validity condition can be
implemented as a Python function.

The function `is_valid()` receives four arguments:

- `board`: the current Sudoku matrix.
- `row`: the zero-based row index.
- `col`: the zero-based column index.
- `num`: the candidate digit.

The function checks whether the selected cell
is empty and whether the candidate satisfies
the row, column, and subgrid restrictions.

It returns `True` if the move is locally valid
and `False` otherwise.

The function does not modify the Sudoku board.

```python
def is_valid(board, row, col, num):
    if not (0 <= row < 9 and 0 <= col < 9 and 1 <= num <= 9):
        return False

    if board[row][col] != 0:
        return False

    if num in board[row]:
        return False

    for r in range(9):
        if board[r][col] == num:
            return False

    start_row = (row // 3) * 3
    start_col = (col // 3) * 3

    for r in range(start_row, start_row + 3):
        for c in range(start_col, start_col + 3):
            if board[r][c] == num:
                return False

    return True
```

## 6. Recursive Backtracking Algorithm

### 6.1 Mathematical Principle

Backtracking is a systematic search method used to solve problems by constructing a solution step by step. If a partial solution cannot be completed, the algorithm returns to a previous step and tries another possibility.

Sudoku can be treated as a constraint satisfaction problem (CSP). Each empty cell is a variable, the possible values are the digits from 1 to 9, and the constraints require that no digit is repeated in any row, column, or 3 × 3 subgrid [1, 2].

Let $A$ represent the Sudoku matrix and let $E(A)$ be the number of empty cells.

A complete solution must satisfy:

$$
E(A)=0
$$

For an empty cell $(i,j)$, the set of possible values is:

$$
C_{i,j}=D\setminus(R_i\cup K_j\cup B_{i,j})
$$

where:

- $D=\{1,2,\ldots,9\}$ is the set of allowed digits.
- $R_i$ contains the nonzero digits in row $i$.
- $K_j$ contains the nonzero digits in column $j$.
- $B_{i,j}$ contains the nonzero digits in the corresponding 3 × 3 subgrid.

For each candidate $v\in C_{i,j}$, the algorithm temporarily assigns:

$$
a_{i,j}=v
$$

If the remaining puzzle cannot be solved, the assignment is reversed:

$$
a_{i,j}=0
$$

This process is called backtracking.

The algorithm explores a search tree in which each node represents a partial assignment and each branch represents a candidate digit. The search follows one branch until it finds a solution or reaches a dead end [1, 2].


```python

def find_empty(board):
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:
                return row, col

    return None


def solve_sudoku(board):
    empty_cell = find_empty(board)

    if empty_cell is None:
        return True

    row, col = empty_cell

    for num in range(1, 10):
        if is_valid(board, row, col, num):
            board[row][col] = num

            if solve_sudoku(board):
                return True

            board[row][col] = 0

    return False
```


### 6.2 Experiment 9: Solving a Sudoku Puzzle

The purpose of this experiment is to solve the original Sudoku puzzle using recursive backtracking.

The algorithm searches for empty cells, tests candidate digits, and recursively constructs a complete solution.

A copy of the original matrix is used because the solving function modifies the board directly.

The expected result is a complete Sudoku matrix containing no empty cells and satisfying all row, column, and subgrid constraints.


```python

original_board = [row.copy() for row in board]
solved_board = [row.copy() for row in board]

success = solve_sudoku(solved_board)

print("Solution found:", success)

if success:
    for row in solved_board:
        print(row)
```


**Experiment 9 - Result Analysis**

The recursive backtracking algorithm successfully solves the original Sudoku puzzle.

The resulting matrix contains no empty cells. Every row, column, and 3 × 3 subgrid contains the digits from 1 to 9 exactly once.

The experiment demonstrates how a constraint satisfaction problem can be solved by combining local validity checks with recursive search.



### 6.3 Experiment 10: Validating the Solution

Finding a complete matrix is not sufficient to prove that a Sudoku solution is correct. The result must satisfy all Sudoku constraints.

Let:

$$
D=\{1,2,3,4,5,6,7,8,9\}
$$

For every row $i$, the set of values must equal $D$:

$$
\{a_{i,0},a_{i,1},\ldots,a_{i,8}\}=D
$$

The same condition must hold for every column and every 3 × 3 subgrid.

Additionally, all original clues must remain unchanged.

The following experiment checks both the mathematical validity of the completed matrix and the preservation of the original clues.


```python

def check_solution(board):
    digits = set(range(1, 10))

    for i in range(9):
        if set(board[i]) != digits:
            return False

        column = {board[r][i] for r in range(9)}
        if column != digits:
            return False

    for start_row in range(0, 9, 3):
        for start_col in range(0, 9, 3):
            box = set()

            for r in range(start_row, start_row + 3):
                for c in range(start_col, start_col + 3):
                    box.add(board[r][c])

            if box != digits:
                return False

    return True


def givens_preserved(original, solution):
    for r in range(9):
        for c in range(9):
            if original[r][c] != 0:
                if original[r][c] != solution[r][c]:
                    return False

    return True


print("Valid solution:", check_solution(solved_board))
print("Original clues preserved:",
      givens_preserved(original_board, solved_board))
```


### 6.4 Mathematical Interpretation and Summary

Recursive backtracking solves Sudoku by exploring possible assignments in a search tree.

At each step, the algorithm selects an empty cell and attempts to assign a digit that satisfies the Sudoku constraints. If a recursive attempt fails, the algorithm removes the previous assignment and tries another candidate.

If the initial puzzle contains $m$ empty cells, the number of possible complete assignments is at most:

$$
9^m
$$

This is a loose upper bound because the Sudoku constraints eliminate many invalid assignments during the search.

The recursion terminates because each recursive call assigns a value to an empty cell, reducing the number of remaining empty cells. If no candidate leads to a solution, the algorithm returns to the previous state.

The solver was applied to a Sudoku puzzle containing 51 empty cells. It produced a complete solution, and additional validation confirmed that the Sudoku constraints were satisfied and the original clues were preserved.

This experiment demonstrates the practical application of recursion, sets, constraints, and systematic search in solving a mathematical puzzle [1, 2, 3].

## 8. Computational Complexity of Backtracking

### 8.1 Theoretical Time Complexity

The recursive backtracking algorithm solves Sudoku by assigning candidate digits to empty cells and checking whether each assignment satisfies the Sudoku constraints.

Let $m$ be the number of initially empty cells in the puzzle.

Each empty cell can potentially contain one of nine digits:

$$
D = \{1,2,3,4,5,6,7,8,9\}
$$

If all nine values were considered independently for each empty cell, the total number of possible complete assignments would be:

$$
N = 9^m
$$

For the Sudoku puzzle used in this project:

$$
m = 51
$$

Therefore:

$$
N = 9^{51} \approx 4.64 \times 10^{48}
$$

This is a theoretical upper bound on the number of complete candidate assignments, not the number of operations performed by the program.

The backtracking search tree has at most nine branches at each level and a maximum depth of $m$.

Consequently, the number of search-tree nodes is bounded by:

$$
1 + 9 + 9^2 + \cdots + 9^m
= \frac{9^{m+1}-1}{8}
$$

The number of nodes is therefore $O(9^m)$.

Since the Sudoku board has a fixed size of 9 × 9, checking whether a digit is valid requires a bounded number of operations. Thus, the search has an exponential worst-case upper bound:

$$
T(m) = O(9^m)
$$

In practice, Sudoku constraints eliminate many invalid candidates before the search reaches a complete assignment.

Backtracking also stops exploring a branch when an assignment violates the constraints.

As a result, the actual number of explored states can be much smaller than the theoretical upper bound.

The algorithm requires additional memory for recursive calls. The recursion depth is at most $m$, so the auxiliary recursion space is:

$$
S(m) = O(m)
$$

These bounds describe the search in terms of the number of empty cells. For a standard 9 × 9 Sudoku, the board size and $m \leq 81$ are fixed.

The analysis follows the general principles of backtracking and search-tree complexity discussed in [2] and [3].
