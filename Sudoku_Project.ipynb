## 3. Mathematical Background

### 3.1 Sudoku as a Constraint Satisfaction Problem

Sudoku is a mathematical and logical puzzle that can be modeled as a Constraint Satisfaction Problem (CSP).

A standard Sudoku board consists of 81 cells arranged in a 9 x 9 grid, divided into nine 3 x 3 subgrids [1].

The goal is to fill the empty cells with digits from 1 to 9 while satisfying the following constraints:

1. Each row must contain every digit from 1 to 9 exactly once.
2. Each column must contain every digit from 1 to 9 exactly once.
3. Each 3 x 3 subgrid must contain every digit from 1 to 9 exactly once.
4. The original given digits must remain unchanged.

**Mathematical Model**

A Constraint Satisfaction Problem consists of three main components: variables, domains, and constraints [2].

**1. Variables**

Each cell of the Sudoku board is represented by a variable:

$$
X_{r,c}
$$

where \(r\) represents the row and \(c\) represents the column.

A standard Sudoku puzzle contains 81 such variables.

**2. Domains**

The domain of a variable represents the possible values that can be assigned to it.

For an initially empty cell, the domain is:

$$
D_{r,c} = \{1,2,3,4,5,6,7,8,9\}
$$

For a cell containing an original given digit, the domain contains only that fixed value.

**3. Constraints**

The mathematical constraints determine which combinations of values are allowed.

All variables belonging to the same row must have different values. The same condition applies to columns and 3 x 3 subgrids.

For example, two different cells in the same row must satisfy:

$$
X_{r,c_1} \neq X_{r,c_2}, \quad c_1 \neq c_2
$$

Similar conditions apply to cells belonging to the same column or subgrid.

Peter Norvig [1] describes Sudoku using units and peers. A unit is a row, column, or 3 x 3 subgrid. Peers are other cells that share a unit with a given cell.

**Connection to the Solving Algorithm**

The objective is to assign a valid value to every empty cell while satisfying all mathematical constraints simultaneously.

The backtracking algorithm will use these constraints to reject invalid assignments and search for a complete solution.

This mathematical model provides the foundation for the Sudoku-solving algorithm developed in this project.

### 3.2 Set Theory and Missing Digits

Set theory provides a mathematical way to represent the allowed and missing digits in Sudoku.

The set of allowed digits is:

$$
D = \{1,2,3,4,5,6,7,8,9\}
$$

Let \(R\) represent the set of digits already used in a row.

The missing digits can be found using set difference:

$$
M = D \setminus R
$$

For example, consider the following row:

`5 3 0 0 7 8 9 1 2`

The set of used digits is:

$$
R = \{1,2,3,5,7,8,9\}
$$

Therefore:

$$
M = D \setminus R = \{4,6\}
$$

In this project, zero represents an empty cell and is excluded from the set of used digits.

Set difference is a fundamental operation in set theory and discrete mathematics [3].

In Sudoku, this operation allows us to determine which
digits are missing from a particular row.

However, identifying missing digits is not sufficient
to determine their exact positions. Each candidate must
also satisfy the column and subgrid constraints [1].


**Experiment 1: Finding Missing Digits**

The purpose of this experiment is to identify missing digits in a partially completed Sudoku row.

First, we create the set of all allowed digits. Next, we convert the row into a set and remove zero, because zero represents an empty cell.

Finally, we subtract the used digits from the allowed digits and display the result in ascending order.

The experiment demonstrates the mathematical concept of
set difference described in discrete mathematics [3].

```python

digits = set(range(1, 10))
row = [5, 3, 0, 0, 7, 8, 9, 1, 2]
used_numbers = set(row) - {0}
missing_numbers = digits - used_numbers

print("Allowed digits:", sorted(digits))
print("Used digits:", sorted(used_numbers))
print("Missing digits:", sorted(missing_numbers))
```

**Experiment 1 - Result Analysis**

The program returns `[4, 6]`, which confirms the mathematical calculation.

The set difference operation successfully identifies the digits that are missing from the row.

However, this result does not determine the exact position of each digit. Both values must also satisfy the column and 3 × 3 subgrid constraints.

This experiment demonstrates that set operations can reduce the possible values, but additional checks are necessary to solve the puzzle.

### 3.3 Permutations and Factorials

Combinatorics is a branch of mathematics that studies
counting and arrangements.

A permutation is an arrangement of distinct elements
in which the order matters [3].

The number of permutations of $n$ distinct elements is:

$$
P_n = n!
$$

For three digits, the number of permutations is:

$$
3! = 3 \times 2 \times 1 = 6
$$

A completed Sudoku row contains nine distinct digits.
Therefore, the number of possible arrangements of
a single row is:

$$
9! = 362880
$$

This result represents the number of possible
permutations of the digits from 1 to 9 in one row.

However, it does not represent the number of valid
solutions for the entire Sudoku puzzle.

The values in each row must also satisfy the
constraints imposed by the columns, the 3 x 3
subgrids, and the original given digits [1].

Norvig [1] explains that each completed Sudoku unit
contains a permutation of the digits from 1 to 9.

In this project, factorial calculations are used
to demonstrate the mathematical concept of
permutations and to illustrate the importance
of constraints in reducing possible arrangements.

The Python function `math.factorial()` is used
to verify the mathematical calculations.

**Experiment 2: Calculating Permutations**

The purpose of this experiment is to calculate the number of possible arrangements of distinct digits.

We first calculate the permutations of three digits and then calculate the permutations of nine digits.

The mathematical concept of factorials is described
in discrete mathematics [3].

The Python `math.factorial()` function is used to
verify the calculated values.

```python

import math
permutations_3 = math.factorial(3)
permutations_9 = math.factorial(9)

print("3! =", permutations_3)
print("9! =", permutations_9)
```

**Experiment 2 - Result Analysis**

The program confirms that three distinct digits have 6 possible permutations, while nine distinct digits have 362,880 permutations.

The result illustrates how rapidly the number of possible arrangements increases.

For Sudoku, permutations alone are not sufficient because the rows, columns, and subgrids must satisfy additional constraints.

This explains why an algorithm should eliminate invalid possibilities rather than blindly consider every arrangement.

### 3.4 Finding Candidate Values

Identifying missing digits in a row is only the first
step in solving Sudoku.

A possible value for an empty cell must satisfy three
constraints simultaneously: it must not already appear
in the corresponding row, column, or 3 x 3 subgrid [1].

Let:

- $D$ be the set of allowed digits.
- $R$ be the set of digits used in the row.
- $K$ be the set of digits used in the column.
- $B$ be the set of digits used in the 3 x 3 subgrid.
- $C$ be the set of possible candidate values.

The candidates can be calculated using:

$$
C = D \setminus (R \cup K \cup B)
$$

This expression combines two fundamental operations
from set theory: union and set difference [3].

The union operation combines the digits already used
in the row, column, and subgrid.

The set difference operation removes these digits
from the complete set of allowed values.

**Mathematical Example**

Consider the following sets:

$$
D = \{1,2,3,4,5,6,7,8,9\}
$$

$$
R = \{3,5,7\}
$$

$$
K = \{8\}
$$

$$
B = \{3,5,6,8,9\}
$$

First, we calculate the union:

$$
R \cup K \cup B = \{3,5,6,7,8,9\}
$$

Then, we calculate the set difference:

$$
C = D \setminus (R \cup K \cup B)
$$

$$
C = \{1,2,4\}
$$

Therefore, the possible candidate values are 1, 2,
and 4.

These values satisfy the local row, column, and
subgrid restrictions.

However, a locally valid candidate does not
necessarily lead to a complete Sudoku solution.

The backtracking algorithm will use these candidate
sets to explore possible assignments and return
to previous decisions when a contradiction occurs.

In Python, set union is implemented using `|`,
and set difference using `-`.

This demonstrates how mathematical set operations
can be translated into Python code.

**Experiment 3: Finding Candidate Values**

In this experiment, we use Python sets to determine the possible values for an empty Sudoku cell.

The sets of used digits in the row, column, and subgrid are provided manually.

We combine these sets using the union operator and subtract the result from the set of allowed digits.

This demonstrates the mathematical idea of eliminating invalid candidates before attempting to solve the puzzle.

```python

digits = set(range(1, 10))
row_values = {3, 5, 7}
column_values = {8}
box_values = {3, 5, 6, 8, 9}

used_numbers = row_values | column_values | box_values
candidates = digits - used_numbers

print("Used digits:", sorted(used_numbers))
print("Candidate values:", sorted(candidates))
```

**Experiment 3 - Result Analysis**

The program returns three possible candidate values: `[1, 2, 4]`.

These values are not present in the corresponding row, column, or subgrid.

The experiment confirms that set operations can be used to eliminate values that violate the Sudoku constraints.

However, a locally valid candidate may still lead to an unsuccessful solution later.

This is one reason why backtracking is useful: the algorithm can try a candidate, continue solving, and return to an earlier decision if a contradiction occurs [1].
